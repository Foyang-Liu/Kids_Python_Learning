## Day 50：类和对象的基本概念

**学习目标**：理解\"类\"和\"对象\"的关系（类是设计图/模板，对象是根据图纸制造出来的具体实例），会用class定义一个简单的类并创建对象

### 📖 知识点讲解

**讲解要点**

- **`class`关键字用来定义一个"类"**，比如 `class Dog:`，类就像一张"设计图"或"模板"，它本身不是一个具体的东西，而是描述了"这一类东西应该是什么样子"。
  **实际场景**：游戏里的"敌人"类型就是一个模板（设计图），游戏运行时会根据这个模板创建出很多个具体的敌人角色，每个敌人都有自己的位置和血量，但都遵循"敌人"这个模板定义好的规则。

  **实际场景**：工厂里的"汽车设计图"是一个类，工厂造出来的每一台具体的汽车都是一个对象——它们外形一样，但车牌号、里程数各不相同。

## Day 51：`__init__`方法——给对象设置初始属性

**学习目标**：掌握__init__方法的写法和作用，理解self参数，会给对象定义属性

### 📖 知识点讲解

**讲解要点**

- **`__init__`方法会在创建对象的那一刻自动被调用**，专门用来给这个新对象设置初始的属性（数据），比如：
  ```python
  class Dog:
      def __init__(self, name, age):
          self.name = name
          self.age = age
  ```
  **实际场景**：新建一个游戏角色时，系统通常会先问你"起什么名字""选什么职业"，然后把这些信息存进这个角色的数据里——这个"创建时自动询问并存储初始信息"的过程，就是`__init__`在做的事。

- **`self`代表"这个对象自己"**，必须是每个方法的第一个参数（调用的时候不需要自己传，Python会自动把"这个对象"传进去）。
  **实际场景**：就像你说"我的书包""我的作业"的时候，"我的"指的就是你自己这个具体的人；`self`在类里面的作用也是一样，指的是"当前这一个具体的对象"，而不是这个类的其他对象。

- **用`self.属性名 = 值`设置的属性，之后可以通过`对象.属性名`从外面访问**。
  **实际场景**：新建一个联系人时填写的姓名、电话，就相当于这个"联系人对象"的属性，存起来之后随时可以通过"打开这个联系人"查看到。

## Day 52：实例方法——给类定义行为

**学习目标**：掌握如何给类定义方法，理解方法如何访问和修改对象自身的属性

### 📖 知识点讲解

**讲解要点**

- **方法就是定义在类里面的函数**，第一个参数也是`self`，通过`self.属性名`可以在方法内部访问或修改"这个对象自己"的属性。
  **实际场景**：游戏角色的"攻击"动作就是一个方法，调用时会根据角色自己的攻击力属性去计算造成的伤害——不同角色攻击力不同，即使调用的是同一个"攻击"方法，造成的伤害也不一样。

- **调用方法用`对象.方法名()`**，不需要手动传`self`，Python会自动把这个对象本身传进去。
  **实际场景**：当你对一只具体的狗说"叫一声"的时候，不需要额外说明"是哪只狗"，因为你已经在对着这只具体的狗发出指令了——`狗对象.叫()`就是这种"对着具体某个对象发出指令"的感觉。

- **方法内部修改的`self`属性，调用之后这个变化会一直保留**，下次再访问这个对象的属性时能看到最新的值。
  **实际场景**：手机的"充电"操作就像是一个方法，调用之后"电量"这个属性会增加，而且这个变化会一直保留，直到你再调用别的方法把它改变。

## Day 55：综合实战——为Checkpoint设计Pet类的基础结构

**学习目标**：练习把前几天学的知识点整合起来，为明天的Checkpoint打基础（今天先搭建基础属性，并尝试加一个带\"边界检查\"的方法）

### 📖 知识点讲解

**讲解要点**

- **设计一个类之前，先想清楚这个类应该有哪些属性（数据）和哪些方法（行为）**，比如"宠物"应该有"名字""饥饿值""心情值"这些属性，应该有"喂食""玩耍"这些方法——这个"先列清单再动手写代码"的过程，是面向对象设计的第一步。
  **实际场景**：做游戏开发时，策划在设计一个新角色之前，通常会先列出这个角色需要具备哪些数值和技能，这个"列清单"的过程就是面向对象设计思路的雏形。
