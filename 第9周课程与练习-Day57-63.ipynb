## Day 57：继承的基本概念

**学习目标**：理解\"继承\"的意义（子类可以自动获得父类已经写好的属性和方法），会用class 子类名(父类名): 的语法定义一个子类

### 📖 知识点讲解

**讲解要点**

- **继承让一个新类可以"直接拿到"另一个类已经写好的东西**，写法是 `class Dog(Animal):`，这样`Dog`就自动拥有了`Animal`里定义的属性和方法，不需要重新写一遍。被继承的类叫"父类"，新定义的类叫"子类"。
  **实际场景**：手机里"智能手机"是从"手机"这个更基础的类型继承来的——它自动拥有"手机"该有的所有基本功能（打电话、发短信），然后再加上自己特有的功能（拍照、上网），不需要把"打电话"这种基础功能重新发明一遍。

- **如果子类什么都不额外写（类的内容只写`pass`），它和父类的行为是完全一样的**，只是多了一个新的"名字"（类型）。
  **实际场景**：公司里"实习生"和"正式员工"都属于"员工"这个大类，如果实习生暂时没有任何特殊规定，那他享有的待遇和"员工"这个基础身份是完全一样的，只是多了一个"实习生"的标签。

- **可以用`isinstance(对象, 类名)`检查一个对象是不是某个类（或者它的父类）的实例**，子类对象用这个函数去检查父类时，结果也会是`True`。
  **实际场景**：检查一只"金毛"是不是"狗"，答案显然是"是"，因为"金毛"这个具体品种本身就是"狗"这个大类别的一种——哪怕问得更笼统一点"这是不是一个动物"，答案也还是"是"。

## Day 58：覆写（Override）父类的方法

**学习目标**：理解子类可以重新定义一个和父类同名的方法，调用时会优先使用子类自己的版本

### 📖 知识点讲解

**讲解要点**

- **子类可以重新写一个和父类里"同名"的方法，这叫"覆写"（override）**，覆写之后，子类对象调用这个方法时，用的是子类自己写的新版本，而不是父类原来的版本。
  **实际场景**：很多游戏里"精英怪"和普通怪物都继承自同一个"怪物"类，都有"攻击"这个动作，但精英怪往往会"覆写"攻击方法，换成更强、更特殊的攻击效果——同一个动作名字，不同的具体表现。

- **覆写不会影响父类本身，也不会影响其他没有覆写的子类**——父类的原始版本还在，只是这一个子类选择了"用自己的版本替换"。
  **实际场景**：不同品牌的手机都继承了"智能手机"这个基础类型，都有"拍照"这个功能，但每个品牌对"拍照"这个功能的具体实现可能完全不同——这些品牌各自"覆写"了拍照功能，但并不会影响其他品牌的实现。

## Day 59：用`super()`调用父类的方法，并给子类添加新内容

**学习目标**：掌握在子类的__init__里用super().__init__(...)调用父类的初始化逻辑，并理解子类可以拥有父类没有的全新属性和方法

### 📖 知识点讲解

**讲解要点**

- **子类的`__init__`里可以用`super().__init__(参数)`，让父类先把它原本该做的初始化工作做一遍**，然后子类再接着设置自己独有的新属性，这样不需要把父类`__init__`里的代码重复抄一遍。
  ```python
  class Animal:
      def __init__(self, name):
          self.name = name

  class Dog(Animal):
      def __init__(self, name, breed):
          super().__init__(name)   # 让Animal先把name设置好
          self.breed = breed       # Dog自己再加一个新属性
  ```
  **实际场景**：办理"学生银行卡"的时候，银行通常会先走一遍"普通银行卡"都要走的基础开户流程，然后再额外加上"学生卡"专属的那部分设置——基础流程交给"父类"去做，自己只需要补充"多出来的那部分"。

- **子类不仅可以继承父类的东西，还可以给自己定义全新的属性和方法，这些新内容父类是没有的**。
  **实际场景**："智能手表"继承了"手表"的基础功能（看时间），但又自己加上了父类"手表"完全不具备的新功能（测心率、连接APP）——这些新功能只属于子类，普通手表没有也用不了。

## Day 60：多态——同一个方法名，不同对象不同表现

**学习目标**：理解\"多态\"的概念：不同的对象只要有同名的方法，就可以被用统一的方式调用，但各自表现出不同的行为

### 📖 知识点讲解

**讲解要点**

- **多态指的是：不同的类，只要定义了同名的方法（通常是各自覆写同一个父类的方法），就可以把它们放在一起，用完全相同的方式去调用，而每个对象表现出来的具体行为可以不一样**。
  **实际场景**：游戏里的"移动"按钮对所有角色都是同一个按钮，但是"人类角色"按下移动是"走"，"飞行角色"按下移动是"飞"，"水中角色"按下移动是"游"——调用的是同一个"移动"指令，不同角色表现出完全不同的动作，这就是多态最直观的体现。

- **多态的好处是写代码的人不需要为每一种具体类型单独写一遍调用逻辑**，只要大家都有同名的方法，直接统一循环调用就行，新增一种新的类型时，原来的调用代码完全不用改。
  **实际场景**：遥控器上的"开/关"按钮可以用来控制电视、空调、风扇——每一种设备收到"开/关"信号后具体做的事情不一样，但使用遥控器的人不需要为每种设备单独学一套操作方式。

## Day 61：json基础——字典与字符串的互相转换

**学习目标**：理解json是什么、为什么要用它，掌握json.dumps和json.loads的用法

### 📖 知识点讲解

**讲解要点**

- **json是一种用文字表示数据的通用格式，长得很像Python的字典**，很多程序、网站之间传递数据都用json格式，因为几乎所有编程语言都能读懂它，不只是Python。
  **实际场景**：手机上不同的App经常需要把数据发给服务器，或者从服务器接收数据，这些数据在传输过程中，很多时候就是用json这种格式包装的——它是不同程序之间"通用的交流语言"。

- **`json.dumps(字典)`可以把一个Python字典转换成一个json格式的字符串**，方便存储或者传输；反过来，**`json.loads(字符串)`可以把一个json格式的字符串转换回Python字典**。

- **包含中文的字典转成json字符串时，加上`ensure_ascii=False`参数，中文才会正常显示，而不是变成一串看不懂的编码**。
  **实际场景**：很多App存储用户资料（比如昵称里有中文）的时候，如果不加这个设置，存出来的文件打开后中文会变成一堆`\u4e2d\u6587`这样的乱码，所以实际写代码时通常都会把这个参数设置好。

## Day 62：把json数据存到文件里，再读回来

**学习目标**：掌握用json.dump和json.load把数据写入文件、从文件读出来，并处理文件不存在的情况

### 📖 知识点讲解

**讲解要点**

- **`json.dump(数据, 文件对象)`可以直接把数据写进一个打开的文件里**，**`json.load(文件对象)`可以直接从一个打开的文件里读出数据并还原**——注意它们和`json.dumps`/`json.loads`（带`s`的那两个，操作的是字符串）不是同一个函数，用法上不要搞混。
  ```python
  with open("data.json", "w", encoding="utf-8") as f:
      json.dump(我的数据, f, ensure_ascii=False)

  with open("data.json", "r", encoding="utf-8") as f:
      还原的数据 = json.load(f)
  ```
