## Day 64：pygame窗口基础

### 📖 知识点讲解

**讲解要点**

- **pygame是一个专门用来做2D游戏的Python工具库**，它能帮你打开一个窗口、画图形、播放声音、检测键盘和鼠标操作——前面9周学的都是"文字版"程序，从这周开始，终于要做出真正"看得见、能操作"的图形界面了。
  **实际场景**：市面上很多经典小游戏（贪吃蛇、打砖块）最早的版本都是用类似pygame这样的简单图形库做出来的——它不像大型游戏引擎那么复杂，但足够做出一个完整能玩的小游戏。

## Day 65：游戏主循环与事件处理

**学习目标**：理解游戏主循环的概念，会用pygame.event.get()获取事件列表，理解怎么判断用户是否点击了关闭按钮

### 📖 知识点讲解

**讲解要点**

- **一个pygame游戏的核心是一个不停重复运行的`while`循环，叫"游戏主循环"**，每一轮循环通常做三件事：①处理这一轮发生了哪些事件 ②更新游戏状态 ③把最新的画面显示出来。这个循环每秒会跑很多次，所以画面看起来是"动"的。
  **实际场景**：动画片是由一张一张静止的图片快速播放组成的，播放速度够快，人眼就会觉得是连续的动作——游戏主循环每一轮画一次"当前画面"，循环跑得够快，看起来就是一个流畅运行的游戏。

- **`pygame.event.get()`会返回"从上一次调用到现在，发生过的所有事件"组成的一个列表**，比如用户按了键盘、点击了窗口的关闭按钮，这些都会变成列表里的一个个"事件对象"。
  **实际场景**：手机的"通知中心"会把你离开手机这段时间收到的所有消息都列出来——`pygame.event.get()`就像是"打开通知中心"，一次性拿到这段时间里发生的所有事情。

- **每个事件对象都有一个`.type`属性，表示这是什么类型的事件**，比如`pygame.QUIT`表示"用户点击了窗口的关闭按钮"，`pygame.KEYDOWN`表示"用户按下了某个键"。判断用户是否想关闭游戏，通常就是在事件列表里找有没有`type`等于`pygame.QUIT`的事件。
  **实际场景**：快递柜的提示音有很多种，每种提示音代表一种不同类型的"事件"，你需要先分辨是哪种提示音（类型），才知道接下来该做什么。

**学习目标**：理解怎么检测哪些键被按下，掌握方向键对应的常量名

### 📖 知识点讲解

**讲解要点**

- **pygame里，上下左右方向键对应的常量分别是`pygame.K_UP`、`pygame.K_DOWN`、`pygame.K_LEFT`、`pygame.K_RIGHT`**，这些常量本质上就是一个数字，但是用有意义的名字代替数字，代码更容易读懂。
  **实际场景**：遥控器上每个按键背后对应一个具体的红外信号编码，但我们不会去记那些编码数字，而是直接看按键上印的名字——`pygame.K_LEFT`这种常量名，就是给"某个按键"起的一个好记的名字。

- **`pygame.key.get_pressed()`会返回"当前这一刻，所有按键是否被按住"的状态**，可以用`按键状态[pygame.K_LEFT]`这样的方式去查某一个具体的键现在是不是被按住。这跟`KEYDOWN`事件不太一样：`KEYDOWN`事件只在"刚按下的那一瞬间"触发一次，而`get_pressed()`可以用来判断"现在是不是一直按着"（更适合用来控制"持续移动"）。
  **实际场景**：开车时，"方向灯刚被打开的那一下"和"方向灯现在是不是亮着"是两种不同的问题——前者像`KEYDOWN`事件，后者像"持续按住的状态"。

## Day 69：边界检测——不能移出窗口

**学习目标**：理解为什么需要限制坐标的范围，掌握怎么用min/max把坐标卡在窗口范围内

### 📖 知识点讲解

**讲解要点**

- **如果不做任何限制，物体的坐标可以一直变大或变小，很快就会移出窗口看不见了**，所以需要在每次更新坐标之后，检查一下是不是超出了窗口的范围，如果超出了，就把坐标"拉回"到最边缘的位置——这跟第8周讲过的"属性边界检查"是同一个思路，只是这次限制的是坐标。
  **实际场景**：手机屏幕上下滑动看消息列表时，滑到最顶或者最底，内容不会继续往外跑，而是会"弹"一下停在边界——这背后也是一种边界检测。

- **具体做法是：物体左边不能小于0，右边（`x + 宽度`）不能超过窗口宽度；上边不能小于0，下边（`y + 高度`）不能超过窗口高度**。可以用`max(0, 一个数)`保证不小于0，用`min(一个数, 上限)`保证不超过上限，两者结合（`max(0, min(数, 上限))`）就能把数字"卡"在一个范围内，跟第8周限制饥饿值、心情值在0-100之间的写法是同一个技巧。
  **实际场景**：游戏摇杆的输入值通常被限制在-1到1之间，不管你把摇杆推得多用力，数值也不会超过这个范围。
