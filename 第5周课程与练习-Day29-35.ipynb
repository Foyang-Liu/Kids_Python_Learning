### 📖 知识点讲解

  **实际场景**：网购时把购物车里的某件商品删掉；通讯录里删除一个不再联系的人。

## Day 30：列表方法进阶（2）—— pop / index / count，以及列表复制的陷阱

**学习目标**：掌握 pop()、index()、count() 的用法；理解直接赋值和复制列表的区别，避开新手最容易踩的坑

### 📖 知识点讲解

**讲解要点**

- **`list.pop()`**：删除并"返回"列表的最后一个元素。和`remove()`的区别是：`pop()`按"位置"删，`remove()`按"值"删；而且`pop()`会把删掉的元素返回给你。
  **实际场景**：很多软件的"撤销"功能，就是把最后一步操作从历史记录列表里`pop`出来，恢复到上一步的状态。

- **`list.index(值)`**：查找某个值在列表中第一次出现的位置（下标）。**`list.count(值)`**：统计某个值在列表中一共出现了多少次。
  **实际场景**：统计全班成绩单里某个具体分数出现了几次；统计一局游戏里某种道具一共掉落了多少个。

- **列表复制的陷阱：`b = a` 和 `b = a.copy()` 的区别**：`b = a` 并不是复制出一份新列表，而是让`b`和`a`成为同一个列表的两个"名字"，改动`b`会直接影响`a`；只有用`a.copy()`（或`a[:]`）才是真正复制出一份独立的新列表。
  **实际场景**：这是新手写游戏存档、多人共享数据时最容易踩的坑——以为复制了一份名单单独修改，结果原始名单也被意外改动了。
