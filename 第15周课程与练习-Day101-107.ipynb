### 📖 知识点讲解

**讲解要点**

- **第14周的关卡表只控制了"掉落物下落的速度"，但一个关卡变难，往往不只是"更快"，还可以是"东西出现得更频繁"**。可以给关卡表再加一列"生成间隔"（比如"每隔多少帧生成一个新掉落物"），数字越小，掉落物出现得越密集、游戏就越紧张。

## Day 104：简单敌人AI——让掉落物追踪玩家

**学习目标**：理解最基础的一种游戏AI思路——比较自己和目标的位置，决定该往哪个方向移动

### 📖 知识点讲解

**讲解要点**

- **最简单的"追踪"AI，本质上就是不断问自己一个问题："我在目标的左边还是右边？"**——如果掉落物的x坐标比玩家的x坐标小，就该往右移动一点；比玩家大，就该往左移动一点；刚好相等，就不需要左右移动。这会让玩家感觉"这东西好像在针对我"，游戏的紧张感和互动感都会提升。
  **实际场景**：扫地机器人判断"障碍物在我的左边还是右边"来决定转向哪一边；很多简单手机游戏里敌人靠近玩家的逻辑，用的也是这种基础思路。

- **要注意"追踪"和"下落"是两件可以同时发生、但互不干扰的事**：掉落物该做的事，是"每一帧，y坐标往下掉一点，同时x坐标朝玩家那一点点靠近"——两个方向的移动分别用各自的速度控制，合在一起就是"斜着飘向玩家、同时往下掉"的效果。
  **实际场景**：横版跳跃游戏里角色同时在"水平移动"和"受重力下落"，两个方向的变化各自独立计算，最后加在一起就是实际看到的移动轨迹。

### 📖 知识点讲解

**讲解要点**

- **"粒子特效"听起来很炫酷，但在最基础的层面，一个粒子其实就是一份很简单的数据：它在哪个位置、是什么颜色、还能"活"多久（寿命）**。每一帧，所有粒子的寿命都减少一点，寿命耗尽的粒子就被移除、不再显示——这样就实现了"出现一下，然后自然消失"的效果。
