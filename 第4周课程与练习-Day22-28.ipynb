```python
# 🔧 工具cell：检测练习题用的工具函数，不需要修改，直接运行一次就好
import random
import math

def test_case(func, args=(), kwargs=None, expected=None, desc="", tol=None):
    """调用func(*args, **kwargs)，并把结果跟expected比较，打印✅或❌"""
    kwargs = kwargs or {}
    try:
        result = func(*args, **kwargs)
    except Exception as e:
        print(f"❌ {desc} —— 运行出错：{type(e).__name__}: {e}")
        return False
    if tol is not None:
        try:
            ok = abs(result - expected) <= tol
        except TypeError:
            ok = False
    else:
        ok = result == expected
    if ok:
        print(f"✅ {desc}")
    else:
        print(f"❌ {desc} —— 期望得到 {expected!r}，但实际得到 {result!r}")
    return ok

def test_soft(condition, desc):
    """直接传入一个判断条件（True/False），打印✅或❌"""
    if condition:
        print(f"✅ {desc}")
    else:
        print(f"❌ {desc}")
    return bool(condition)

print("工具函数准备好了，可以开始做练习题了！")
```
