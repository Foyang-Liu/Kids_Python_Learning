```
# 🔧 工具cell：检测练习题用的工具函数，不需要修改，直接运行一次就好
import contextlib
import io
import builtins

class mock_input:
    """检测需要用到input()的练习题时，临时按顺序提供输入内容"""
    def __init__(self, values):
        self._iter = iter(list(values))
        self._original = None
    def __enter__(self):
        self._original = builtins.input
        builtins.input = lambda prompt="": next(self._iter)
        return self
    def __exit__(self, *exc):
        builtins.input = self._original

def _values_match(result, expected, tol):
    if tol is not None:
        try:
            return abs(result - expected) <= tol
        except TypeError:
            return False
    return result == expected

def test_case(func, args=(), kwargs=None, expected=None, desc="", tol=None):
    kwargs = kwargs or {}
    try:
        result = func(*args, **kwargs)
    except Exception as e:
        print(f"❌ {desc} —— 运行出错：{type(e).__name__}: {e}")
        return False
    ok = _values_match(result, expected, tol)
    if ok:
        print(f"✅ {desc}")
    else:
        print(f"❌ {desc} —— 期望得到 {expected!r}，但实际得到 {result!r}")
    return ok

def test_case_with_input(func, inputs, expected=None, desc="", tol=None):
    try:
        with mock_input(inputs):
            result = func()
    except Exception as e:
        print(f"❌ {desc} —— 运行出错：{type(e).__name__}: {e}")
        return False
    ok = _values_match(result, expected, tol)
    if ok:
        print(f"✅ {desc}")
    else:
        print(f"❌ {desc} —— 期望得到 {expected!r}，但实际得到 {result!r}")
    return ok

def test_soft(condition, desc):
    if condition:
        print(f"✅ {desc}")
    else:
        print(f"❌ {desc}")
    return bool(condition)

print("工具函数准备好了，可以开始做练习题了！")
```

## Day 5：算术运算符

**学习目标**：掌握 `+ - * / // % **` 的用法和区别，理解运算优先级

### 📖 知识点讲解

**讲解要点**
- `/` 除法的结果永远是小数（float），比如 `17 / 5` 等于 `3.4`；`//` 是整除，只保留商、不要余数，比如 `17 // 5` 等于 `3`；`%` 是取余数，专门用来求余数，比如 `17 % 5` 等于 `2`。
- `**` 是幂运算，比如 `2 ** 3` 表示2的3次方，结果是8。
- 运算是有优先级的：先乘除后加减，括号里的内容永远最优先计算，这和平时数学课学的顺序是一样的。

## Day 6：比较运算符与 f-string 格式化输出

**学习目标**：掌握6种比较运算符（结果是bool）；学会用f-string让输出更清晰

### 📖 知识点讲解

**讲解要点**
- 六种比较运算符：`==`（等于）、`!=`（不等于）、`<`（小于）、`>`（大于）、`<=`（小于等于）、`>=`（大于等于），它们的结果永远是 `True` 或 `False`。
- 特别提醒：两个等号 `==` 才是比较，一个等号 `=` 是赋值，这是新手最容易搞混的地方，一定要小心区分。
- f-string格式化输出：写成 `f"我叫{name}，今年{age}岁"` 这样的形式，在字符串前面加一个`f`，大括号里直接写变量名，比用逗号一个个拼接要清楚得多，而且还能控制小数点后面保留几位，比如 `f"{pi:.2f}"` 就表示只保留2位小数。
