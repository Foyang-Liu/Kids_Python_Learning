### 📖 知识点讲解

**讲解要点**

- **字典用 `{key: value}` 创建**，通过key直接找到对应的value，比如 `student = {"name": "小明", "age": 13}`。
  **实际场景**：手机通讯录本质上就是"姓名"对应"电话号码"的字典结构；游戏里的角色属性表，比如 `{"生命值": 100, "攻击力": 20}`，比用好几个单独的变量清晰得多，还能随时增加新属性。

- **访问用 `dict[key]`**，如果key不存在会直接报错；更安全的写法是 `dict.get(key)`，key不存在时会返回 `None` 而不是报错。
  **实际场景**：查快递单号对应的物流状态时，如果这个单号根本不存在，用 `get()` 查询程序不会崩溃，只会告诉你"查不到"，这样用户体验更好。

- **修改或新增用 `dict[key] = 新值`**：如果这个key已经存在，就是修改它对应的值；如果不存在，就是新增一条记录。
  **实际场景**：更新通讯录里某个人的电话号码；游戏里角色升级后，把"等级"这个属性对应的值改成新的数字。

**学习目标**：掌握读取文件内容的方法，理解如何应对\"文件不存在\"的情况

### 📖 知识点讲解

**讲解要点**

- **`open(文件名, "r")`**：以"读取模式"打开文件；**`file.read()`** 一次性读出全部内容（一个大字符串）；**`file.readlines()`** 按行读取，返回一个列表，每一行是列表里的一个元素。
  **实际场景**：打开之前保存的游戏存档文件，把里面记录的进度读出来接着玩；重新打开一份之前保存好的通讯录文件，把内容加载回程序里继续使用。

- **推荐用 `with open(文件名, "r") as file:` 这种写法**：Python会自动处理文件的关闭，哪怕中途代码出错也不会忘记关闭文件，这是业界公认更安全的标准写法。
  **实际场景**：几乎所有专业的Python项目在读写文件时都会用 `with` 语句，这是一个很值得从现在开始养成的好习惯。

- **文件不存在时直接读取会报错**，实际项目里通常需要提前处理这种情况（可以配合第4周学过的 `try/except` 异常处理）。
  **实际场景**：第一次打开一个新游戏时，还没有任何存档文件，程序需要能正常处理"文件不存在"这种情况（比如当作"新玩家、从头开始"），而不是直接报错崩溃。
